## Ejercicio 1 — Factorial 

```python
numero = int(input('Ingrese un numero entero: '))
factorial = 1
for i in range(1, numero + 1):
    factorial *= i

print('El factorial de', numero, 'es:', factorial)
```

## Ejercicio 2 — Primo o no

```python
import math
numero = int(input("Ingrese un numero entero: "))

es_primo = True
if numero <= 1:
    es_primo = False
else:
    for i in range(2, numero):
        if numero % i == 0:
            es_primo = False
            break


if es_primo:
    print("El numero", numero, "es primo.")
else:
    print("El numero", numero, "no es primo.")
```

## Ejercicio 3 — Mayor y menor


```python
def mayor_y_menor(numeros):
    if len(numeros) < 3:
        raise ValueError("Se requieren al menos 3 numeros")
    
    return max(numeros), min(numeros)

entrada = input("Ingrese los numeros separados por espacio (minimo 3): ")
numeros = list(map(float, entrada.split()))
mayor, menor = mayor_y_menor(numeros)


print("\nLista ingresada:", numeros)
print("Numero mayor:", mayor)
print("Numero menor:", menor)
```

## Ejercicio 4 — Palindromo

```python
texto = input("Ingrese una palabra: ")

texto_limpio = texto.replace(" ", "").lower()
if texto_limpio == texto_limpio[::-1]:
    print("La cadena ES un palindromo.")
else:
    print("La cadena NO es un palindromo.")
```

## Ejercicio 5 — Cual numero se repite mas


```python
entrada = input("Ingrese 7 numeros separados por espacio: ")


numeros = list(map(float, entrada.split()))
if len(numeros) != 7:
    print("Debe ingresar exactamente 7 números.")
else:
    max_repeticiones = 0
    numero_mas_repetido = None

    for num in numeros:
        contador = numeros.count(num)

        if contador > max_repeticiones:
            max_repeticiones = contador
            numero_mas_repetido = num

    
    print("\nNúmeros ingresados:", numeros)
    print("Número que más se repite:", numero_mas_repetido)
    print("Cantidad de repeticiones:", max_repeticiones)
```

## Ejercicio 6 — Anagrama

```python
palabra1 = input("Ingrese la primera palabra: ")
palabra2 = input("Ingrese la segunda palabra: ")

palabra1 = palabra1.lower()
palabra2 = palabra2.lower()
if sorted(palabra1) == sorted(palabra2):
    print("Las palabras SON anagramas.")
else:
    print("Las palabras NO son anagramas.")
```

## Ejercicio 7 — Metodo Burbuja

```python
entrada = input("Ingrese los numeros separados por espacio: ")

numeros = list(map(int, entrada.split()))
n = len(numeros)

for i in range(n - 1):
    for j in range(n - i - 1):
        if numeros[j] > numeros[j + 1]:
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]


print("Números ordenados:", numeros)
```
