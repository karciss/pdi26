## Ejercicio 1 — Factorial 

```python
# Versión robusta: intenta leer desde input(). Si no hay entrada disponible, usa un valor por defecto para pruebas.
try:
    numero = int(input('Ingrese un numero entero: '))
except (EOFError, KeyboardInterrupt):
    # Entornos sin stdin (ej. algunos runners) lanzan EOFError; usamos un valor por defecto para demostrar la salida.
    numero = 5
    print('No se recibió entrada interactiva; usando numero =', numero)
except Exception as e:
    print('Entrada inválida, usando valor por defecto 5. Error:', e)
    numero = 5

factorial = 1
for i in range(1, numero + 1):
    factorial *= i

print('El factorial de', numero, 'es:', factorial)
```

## Ejercicio 2 — Primo o no

```python
import math
def es_primo(n):
    if not isinstance(n, int):
        raise TypeError('El número debe ser un entero')
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0:
        return False
    limite = int(math.sqrt(n))
    i = 3
    while i <= limite:
        if n % i == 0:
            return False
        i += 2
    return True

# Ejemplos
print('7 es primo?', es_primo(7))
print('12 es primo?', es_primo(12))
print('1 es primo?', es_primo(1))
```

## Ejercicio 3 — Mayor y menor


```python
def mayor_y_menor(numeros):
    if len(numeros) < 3:
        raise ValueError('Se requieren al menos 3 números')
    return max(numeros), min(numeros)

# Ejemplo:
ejemplo = [3, 10, -2, 7, 5]
print('Lista ejemplo:', ejemplo)
print('Mayor, Menor =', mayor_y_menor(ejemplo))
```

## Ejercicio 4 — Palindromo

```python
def es_palindromo(s):
    texto = ''.join(ch.lower() for ch in s if ch.isalnum())
    return texto == texto[::-1]

# Ejemplos
print(es_palindromo('Anita lava la tina'))
print(es_palindromo('hola'))
```

## Ejercicio 5 — Introducir 7 números y determinar cuál se repite más veces y cuántas


```python
from collections import Counter
def mas_frecuente(nums):
    if len(nums) != 7:
        raise ValueError('Se requieren exactamente 7 números')
    c = Counter(nums)
    max_freq = max(c.values())
    mas = [val for val, cnt in c.items() if cnt == max_freq]
    return mas, max_freq

# Ejemplo:
nums = [1,2,2,3,4,2,5]
print(mas_frecuente(nums))
```

## Ejercicio 6 — Identificar anagrama
Identificar si una palabra es un anagrama de otra (ej. Bolo - Lobo).

```python
from collections import Counter
def es_anagrama(a, b):
    a_clean = ''.join(ch.lower() for ch in a if ch.isalnum())
    b_clean = ''.join(ch.lower() for ch in b if ch.isalnum())
    return Counter(a_clean) == Counter(b_clean)

print(es_anagrama('Bolo','Lobo'))
print(es_anagrama('Hola','Aloh'))
```

## Ejercicio 7 — Ordenar números con método burbuja
Desarrolla una aplicación que ordene números a través del método burbuja.

```python
def burbuja(arr):
    a = arr.copy()
    n = len(a)
    for i in range(n):
        for j in range(0, n-i-1):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
    return a

lista = [64, 34, 25, 12, 22, 11, 90]
print('Antes:', lista)
print('Después:', burbuja(lista))
```
