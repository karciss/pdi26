```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# aplicar convoluciones
def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((1,tkernel),np.float32) / tkernel
    imagen_c=cv2.filter2D(imagen,-1,kernel)
    return imagen_c

img=cv2.imread('img.jpg')
imagen=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
img_fin=convolucion_media_h(imagen,35)

def convolucion_media_di(imagen, tkernel):
    kernel = np.fliplr(np.eye(tkernel, dtype=np.float32)) / tkernel
    imagen_c=cv2.filter2D(imagen,-1,kernel)
    return imagen_c

img_fin_di=convolucion_media_di(img_fin,25)

#efecto de nitidez 
def efecto_nitidez(imagen, tkernel):
    kernel = np.array([[0, -1, 0], [-1, tkernel, -1], [0, -1, 0]], dtype=np.float32)
    imagen_c=cv2.filter2D(imagen,-1,kernel)
    return imagen_c

img_fin_nitidez=efecto_nitidez(img_fin_di,5)

#mostrar 
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(imagen,cmap='gray')
plt.title('Original')
plt.subplot(1,2,2)
plt.imshow(img_fin_nitidez,cmap='gray')
plt.title('Imagen Tratada')
plt.show()
```
