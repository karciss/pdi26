## CONVOLUCION DE PRACTICA (ESCALA DE GRISES)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('imagen1.jpg', 0)

filtros = {
    "Original": None,
    
    "Enfoque": np.array([[0,-1,0],
                         [-1,5,-1],
                         [0,-1,0]]),

    "Desenfoque": np.array([[1,1,1],
                            [1,1,1],
                            [1,1,1]]) / 4,

    "Realce bordes": np.array([[0,0,0],
                               [-1,1,0],
                               [0,0,0]]),

    "Repujado": np.array([[-2,-1,0],
                          [-1,1,1],
                          [0,1,2]]),

    "Deteccion de Bordes": np.array([[0,1,0],
                                     [1,-4,1],
                                     [0,1,0]]),

    "Filtro de tipo Sobel X": np.array([[-1,0,1],
                         [-2,0,2],
                         [-1,0,1]]),

    "Filtro de tipo Sobel Y": np.array([[-1,-2,-1],
                         [0,0,0],
                         [1,2,1]]),

    "Sharpen": np.array([[1,-2,1],
                         [-2,5,-2],
                         [1,-2,1]]),

    "Norte": np.array([[1,1,1],
                       [1,-2,1],
                       [-1,-1,-1]]),

    "Este": np.array([[-1,1,1],
                      [-1,-2,1],
                      [-1,1,1]]),

    "Gauss": np.array([[1,2,3,2,1],
                       [2,7,11,7,2],
                       [3,11,17,11,3],
                       [2,7,11,7,2],
                       [1,2,3,2,1]]) / 115
}

plt.figure(figsize=(15,10))

for i, (nombre, kernel) in enumerate(filtros.items()):
    plt.subplot(4, 4, i+1)
    
    if kernel is None:
        resultado = img
    else:
        resultado = cv2.filter2D(img, -1, kernel)
    
    plt.imshow(resultado, cmap='gray')
    plt.title(nombre)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## Horizontal

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_h(imagen,tkernel):
    kernel = np.ones((1,tkernel),np.float32)/tkernel
    imagen_c = cv2.filter2D(imagen,-1,kernel)
    return imagen_c


img = cv2.imread('img2.jpg', 0)

img_fin = convolucion_media_h(img, 13)

plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.title('Original')
plt.subplot(1,2,2)
plt.imshow(img_fin,cmap='gray')
plt.title('Convolucion Media Horizontal')
plt.show()
```

## CONVOLUCION HORIZONTAL DE COLORES

## CONVOLUCION EN COLORES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

#Convoluconal horizontal
def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((1,tkernel),np.float32) / tkernel
    imagen_c=cv2.filter2D(imagen,-1,kernel)
    return imagen_c

#Convoluconal vertical 
def convolucion_media_v(imagen, tkernel):
    kernel = np.ones((tkernel,1),np.float32) / tkernel
    imagen_c=cv2.filter2D(imagen,-1,kernel)
    return imagen_c

#Convoluconal Diagonal
def convolucion_media_d(imagen, tkernel):
    kernel = np.eye(tkernel, dtype=np.float32) / tkernel
    imagen_c=cv2.filter2D(imagen,-1,kernel)
    return imagen_c

#Convoluconal Diagonal Inversa
def convolucion_media_di(imagen, tkernel):
    kernel = np.fliplr(np.eye(tkernel, dtype=np.float32)) / tkernel
    imagen_c=cv2.filter2D(imagen,-1,kernel)
    return imagen_c


img=cv2.imread('img3.jpg')


imagen=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
img_fin_h=convolucion_media_h(imagen,15)
img_fin_v=convolucion_media_v(imagen,15)
img_fin_d=convolucion_media_d(imagen,15)
img_fin_di=convolucion_media_di(imagen,15)

#Mostrar resultados
plt.figure(figsize=(15,10))
plt.subplot(2,3,1)
plt.imshow(imagen,cmap='gray')
plt.title('Original')
plt.subplot(2,3,2)
plt.imshow(img_fin_h,cmap='gray')
plt.title('Convolucion Media Horizontal')
plt.subplot(2,3,3)
plt.imshow(img_fin_v,cmap='gray')
plt.title('Convolucion Media Vertical')
plt.subplot(2,3,4)
plt.imshow(img_fin_d,cmap='gray')
plt.title('Convolucion Media Diagonal')
plt.subplot(2,3,5)
plt.imshow(img_fin_di,cmap='gray')
plt.title('Convolucion Media Diagonal Inversa')
plt.tight_layout()

```
