## USO CV2 - Practica

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## Funcion para mostrar imagenes

```python
def mostrar_imagen(img, titulo='Imagen', cmap=None):
    plt.figure(figsize=(10,5))
    plt.imshow(img, cmap=cmap)
    plt.title(titulo)
    plt.axis('off')
    plt.show()
```

# Imagen 1

### Cargar imagen en formato BGR y mostrar en RGB

```python
img1_bgr = cv2.imread('img1.jpg')
img1_rgb = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2RGB)

mostrar_imagen(img1_rgb, 'Imagen 1 - RGB')
```

### Convertir a escala de grises, guardar y mostrar

```python
img1_gray = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2GRAY)
cv2.imwrite('img1_gray.jpg', img1_gray)

cv2.imwrite('img1_bgr.jpg', img1_bgr)
mostrar_imagen(img1_gray, 'Imagen 1 - Escala de Grises', cmap='gray')
```

### Mostrar canales en escala de grises - Modo Subplot 1

```python
plt.rcParams['figure.figsize'] = [15,5]

plt.subplot(141)
plt.title('Imagen Original')
plt.imshow(img1_rgb)
plt.axis('off')

plt.subplot(142)
plt.title('Canal Rojo')
plt.imshow(img1_rgb[:,:,0], cmap='gray')
plt.axis('off')

plt.subplot(143)
plt.title('Canal Verde')
plt.imshow(img1_rgb[:,:,1], cmap='gray')
plt.axis('off')

plt.subplot(144)
plt.title('Canal Azul')
plt.imshow(img1_rgb[:,:,2], cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Mostrar canales en color - Modo Subplot 2

```python
r = img1_rgb[:,:,0]
g = img1_rgb[:,:,1]
b = img1_rgb[:,:,2]


r_img = np.zeros_like(img1_rgb)
r_img[:,:,0] = r

g_img = np.zeros_like(img1_rgb)
g_img[:,:,1] = g

b_img = np.zeros_like(img1_rgb)
b_img[:,:,2] = b


fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].imshow(img1_rgb)
axs[0].set_title('Imagen Original')
axs[0].axis('off')

axs[1].imshow(r_img)
axs[1].set_title('Canal Rojo')
axs[1].axis('off')

axs[2].imshow(g_img)
axs[2].set_title('Canal Verde')
axs[2].axis('off')

axs[3].imshow(b_img)
axs[3].set_title('Canal Azul')
axs[3].axis('off')

plt.tight_layout()
plt.show()
```

### Mostrar imagen en color falso

```python
plt.rcParams['figure.figsize'] = [15,9]

cmaps = ['viridis','plasma','inferno','magma','cividis','hot','cool','spring','summer','winter']

fig = plt.figure()
gs = fig.add_gridspec(3,5)

# Fila 1
ax = fig.add_subplot(gs[0,:])
ax.set_title('Imagen Original')
ax.imshow(img1_rgb)
ax.axis('off')

# Filas 2 y 3
for i, cmap in enumerate(cmaps):
    row = 1 + i//5
    col = i%5
    ax = fig.add_subplot(gs[row, col])
    ax.set_title(cmap)
    ax.imshow(img1_gray, cmap=cmap, vmin=0, vmax=255)
    ax.axis('off')

plt.tight_layout()
plt.show()
```
